```python
import pandas as pd
import numpy
```

```python
car = pd.read_csv("quikr_car.csv")
```

```python
car.head()
```

```python
backup = car.copy()
```

```python
car = car[car['year'].str.isnumeric()]
```

```python
 car = car[car['Price'] != "Ask For Price"]
```

```python
car['Price'] = car['Price'].str.replace(',','').astype(int)
```

```python
car = car[~car['fuel_type'].isna()]
```

```python
car['kms_driven'] = car['kms_driven'].str.split(' ').str.get(0).str.replace(',','')
```

```python
car = car[car['kms_driven'].str.isnumeric()]
```

```python
car['kms_driven'] = car['kms_driven'].astype(int)
```

```python
car['kms_driven'].describe()
```

```python
kms = car['kms_driven'].copy()
```

```python
kms.head()
```

```python
kms_sorted = kms.sort_values()
```

```python
kms_sorted.tail(20)
```

```python
Q1 = kms_sorted.quantile(0.25)
Q3 = kms_sorted.quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

filtered_kms = kms_sorted[(kms_sorted >= lower) & (kms_sorted <= upper)]
```
